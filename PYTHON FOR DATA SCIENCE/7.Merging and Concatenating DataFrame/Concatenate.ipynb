```python
# Concatenate Vertically

import pandas as pd

df_movies = pd.read_csv(r"D:\python-for-data-science-main\09.Merging and Concatenating DataFrames\IMDb movies.csv\IMDb movies.csv", low_memory=False)
df_movies

df_sample = df_movies.sample(frac=0.5)

print(df_movies.shape)
print(df_sample.shape)
```

```python
# Concatenate Horizontally

import pandas as pd

df_movies = pd.read_csv(r"D:\python-for-data-science-main\09.Merging and Concatenating DataFrames\IMDb movies.csv\IMDb movies.csv", low_memory=False)
df_movies

df1 = pd.DataFrame({'id':['A', 'B', 'C', 'D'],
                   'age':[30, 23, 25, 22 ]})

df2 = pd.DataFrame({'job':['Doctor', 'Statistician', 
                           'Accountant', 'Developer']})

pd.concat([df1, df2], axis=1)
```
