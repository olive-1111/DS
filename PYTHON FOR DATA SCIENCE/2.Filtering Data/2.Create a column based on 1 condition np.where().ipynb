```python
import pandas as pd

df_laptops = pd.read_csv(r"D:\python-for-data-science-main\04.Filtering Data\laptop_price.csv", encoding='ISO-8859-1')
df_laptops.head(3)
```

```python
import numpy as np

df_laptops['Price_euros']>2000
```

```python
np.where(df_laptops['Price_euros']>2000, 'Expensive', 'Cheap')
```

```python
df_laptops['Price_tier'] = np.where(df_laptops['Price_euros']>2000, 'Expensive', 'Cheap')
df_laptops
```
