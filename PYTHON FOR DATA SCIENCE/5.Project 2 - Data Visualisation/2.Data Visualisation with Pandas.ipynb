```python
import pandas as pd

df_population_raw = pd.read_csv(r"D:\python-for-data-science-main\07.Project 2 - Data Visualization\population_total.csv")
df_population_raw
```

```python
df_population_raw.dropna(inplace=True)
df_pivot = df_population_raw.pivot(index='year', columns='country', 
                        values='population')
```

```python
df_pivot = df_pivot[['United States', 'India', 'China', 'Indonesia', 'Brazil']]
df_pivot
```

```python
df_pivot.plot(kind='line')
```

```python
df_pivot.plot(kind='line', xlabel='Year', ylabel='Population', title='Population (1955-2020)')
```

```python
df_pivot.index
```

```python
df_pivot_2020 = df_pivot[df_pivot.index.isin([2020])]
df_pivot_2020
```

```python
df_pivot_2020 = df_pivot_2020.T
df_pivot_2020
```

```python
df_pivot_2020.plot(kind='bar', color='orange', 
                   xlabel='Year', ylabel='Population', 
                   title='Population 2020')
```

```python
df_pivot_sample = df_pivot[df_pivot.index.isin([1980, 1990, 2000, 2010, 2020])]
df_pivot_sample
```

```python
df_pivot_sample.plot(kind='bar', xlabel='Year', ylabel='Population')
```
